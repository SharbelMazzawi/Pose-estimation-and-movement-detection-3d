##  🔍 What is DeepLabCut?

[DeepLabCut](https://www.deeplabcut.org/) is an open-source tool for markerless pose estimation of animals (and humans) using deep learning. It allows researchers to accurately label and track body parts in videos with minimal manual effort, and it integrates well with tools like Napari for efficient frame annotation.

##  ⚙️ Installation

To use DeepLabCut and Napari, we first need to set up a dedicated Python environment using Anaconda. This ensures all dependencies are properly managed. We install DeepLabCut along with its GUI and Napari using `conda` and `pip`. Below are the commands used to create the environment and install all required packages.

If you're new to Python, the easiest way to get started is by installing [Anaconda](https://www.anaconda.com/products/distribution), which lets you manage Python environments and packages easily. Once Anaconda is installed, you can create a clean environment and install DeepLabCut and Napari using the commands below. This setup ensures that all dependencies are isolated and compatible.

## 🖥️ Setting Up the Environment and Installing DeepLabCut

To run the installation commands, you need to open the **Anaconda Prompt** (Windows) or a **Terminal** (Mac/Linux). This is where you'll type the following commands, step by step. Make sure you follow each one in order to properly install DeepLabCut, Napari, and all the necessary dependencies.

````python
# Step 1: Create a new conda environment with Python 3.10
conda create -n deeplabcut python=3.10

# Step 2: Activate the environment
conda activate deeplabcut

# Step 3: Install PyTorch (choose the version that matches your system)

# For GPU with CUDA 12.4
pip install torch torchvision --index-url https://download.pytorch.org/whl/cu124

# OR for CPU only
# pip install torch torchvision --index-url https://download.pytorch.org/whl/cpu

# Step 4: Install dependencies
conda install -c conda-forge pytables==3.8.0

# Step 5: Install DeepLabCut with GUI support
pip install "deeplabcut[gui,modelzoo,wandb]"
````

## 🚀 Launching DeepLabCut

Once DeepLabCut is installed, you can launch its graphical interface from the Anaconda Prompt or Terminal. This opens the DLC App, where you can create a new project and start labeling frames using Napari.

## 🔧 Defining Keypoints to Track

Once your project is created, you’ll need to define which body parts (keypoints) to label in your video frames. This is done by editing the `config.yaml` file from the Project Management tab in the GUI.

### ✏️ Steps to Edit the Configuration:

1. **Open the `config.yaml` File**  
   At the top of the main DeepLabCut GUI window, you’ll see the path to your configuration file. Click on **"Edit config.yaml"** to launch the Configuration Editor.

2. **Define Bodyparts**  
   - Locate the `bodyparts` section.  
   - Add each keypoint you want to label. For example, to track a hand and its fingers, you might define:
     - `right_hand_center`
     - `right_finger1`
     - `right_finger2`
     - `right_finger3`
     - `right_finger4`
     - `right_finger5`
   - You can add new rows by right-clicking and choosing **"Insert"**.

3. **Define the Skeleton (Optional but Recommended)**  
   The skeleton defines how body parts are connected, which helps DeepLabCut better understand spatial relationships.

   - Scroll to the `skeleton` section in the config editor.
   - Add pairs of body parts to define connections. For a hand structure, your skeleton could look like this:

     ```yaml
     skeleton:
       - [right_hand_center, right_finger1]
       - [right_hand_center, right_finger2]
       - [right_hand_center, right_finger3]
       - [right_hand_center, right_finger4]
       - [right_hand_center, right_finger5]
     ```

   This tells DeepLabCut that all five fingers are directly connected to the center of the hand.

💡 *Tip: Skeletons help improve model generalization, especially when multiple points are closely spaced.*

4. **Save Your Changes**  
   When you're done, click **"Save"** to return to the main GUI with your changes applied.   
   
5. **example**  
    There is a config.yaml example file in the folder, go check it out!


## 🖼️ Extracting and Labeling Frames with Napari

After defining keypoints and setting up your project, the next step is to extract frames for manual labeling. This is a crucial step—your model learns from these examples, so selecting a diverse and accurate sample is key.

### 🎯 Selecting Frames to Label

DeepLabCut offers two main methods for frame extraction:

1. **Automatic Extraction**  
   - Click **"Extract frames"** in the Project Management tab (the bar in the top of the screen).
   - Choose **Uniform** to extract frames at regular intervals.
   - Or choose **K-means clustering** for more variation between poses.
   - Set the number of frames (e.g., 100–200).
   - Click **"Extract"** to create labeled-data folders containing the extracted frames.

2. **Manual Selection (Optional)**  
   - You can also choose frames manually to capture rare or specific behaviors by browsing through your videos and selecting frames manually from the GUI.

💡 *Tip: It's often best to combine both—use K-means for general variety, then manually extract rare events.*

---

### 🏷️ Labeling Frames with Napari

1. In the DLC GUI (on the bar in the top of the screen), click **"Label Frames"**.
2. Choose a folder inside the `labeled-data` directory and click **"Open"**. (the folder you are chosing is the one that contain the extracted frames from the previous section which you want to label).
3. The **Napari DeepLabCut plugin** window will open.
4. A short tutorial popup may appear—feel free to read or skip.
5. Begin labeling using the **"Add points"** tool.
   - Click each keypoint (e.g., `right_hand_center`, `right_finger1`, etc.) on the frame.
   - Use the slider or arrow keys to move between frames.
   - Label all visible keypoints in each frame.

💡 *If a keypoint is not visible, skip it or use a default position consistently.*

6. Press **Ctrl+S** (or **Cmd+S** on Mac) to save your labels as you go.
7. Repeat for each folder created for your videos.

---

### 🔍 Checking Your Labels

Once you’re done labeling:
- Return to the main DLC GUI.
- Click **"Check Labels"** to preview your annotations.
- This will generate overlaid images in new folders under `labeled-data/`, allowing you to inspect your work.

Accurate labeling is essential! Errors here can cause the model to mislearn later during training.


## 🧠 Training and Evaluating the Neural Network

Once you've labeled enough frames, it's time to train the DeepLabCut model to predict the keypoints in your videos.

---

### 🛠️ Creating the Training Dataset

1. In the GUI, go to the **"Create Training Dataset"** tab.
2. Click **Create Training Dataset** — for most projects, the default settings are ideal.
3. DeepLabCut will prepare the necessary files to train the neural network.

💡 *You can customize model architecture and augmentations later, but the defaults are a great place to start.*

---

### 🚀 Kickstarting the Training Process

1. Go to the **"Train Network"** tab.
2. Set training parameters:
   - **Display iterations/epochs**: How often progress is updated in the terminal.
   - **Maximum Iterations/Epochs**: How many epochs to train the model.
     - PyTorch: 200 epochs is usually enough.
     - TensorFlow: 10,000 iterations for a quick demo.
   - **Snapshots to keep**: Set how often model states are saved.
3. Click **Train Network** to begin training.

✅ *You’ll see real-time logs in your terminal, especially useful if you’re using the PyTorch engine, which also prints evaluation metrics per epoch.*

💻 **Note for CPU-only users**: If training is slow, consider using [Google Colab](https://colab.research.google.com/github/DeepLabCut/DeepLabCut/blob/master/examples/COLAB/COLAB_YOURDATA_TrainNetwork_VideoAnalysis.ipynb).

---

### 📊 Evaluating Your Model

After training:

1. Go to the **"Evaluate Network"** tab.
2. Choose what to evaluate:
   - **Plot Predictions** to visualize model output vs. labels.
   - **Compare Bodyparts** for a full evaluation of tracking accuracy.
3. Click **Evaluate Network**.

Results include:
- A `.csv` file with performance metrics (in `evaluation-results` or `evaluation-results-pytorch`).
- Images showing original labels and predicted points for easy visual inspection.

💡 *To evaluate all snapshots, open the `config.yaml` and set `snapshotindex: all`.*


## 🎥 Video Analysis with DeepLabCut

After training and evaluating your model, the final step is to apply it to actual videos to estimate keypoints across frames.

---

### ▶️ How to Analyze Videos

1. Go to the **"Analyze Videos"** tab in the DeepLabCut GUI.
2. **Select your video format and files**:
   - Supported formats: `.mp4`, `.avi`, `.mov`, `.mkv`
   - Click **"Select Videos"** to choose your file(s).
3. Click **"Analyze"** to start the process.

💡 *The analysis time depends on the resolution and duration of your video. You can monitor progress in the terminal or Anaconda prompt.*

---

### 📂 Reviewing Analysis Results

After the analysis completes, DeepLabCut saves the results in your project folder:

- `.h5` and `.csv` files containing estimated keypoint data.
- `.pickle` or `.metapickle` file with metadata.
- A `plot-poses` subfolder with plotted images showing estimated bodyparts.

---

### 🎞️ Creating a Labeled Video

To visualize the predictions over your original video:

1. Open the **"Create Labeled Video"** tab.
2. Select your analyzed video.
3. Click **"Create Videos"**.

The output will be saved in your video folder with `_labeled` in the name.

✅ Watch this video to visually inspect the tracking accuracy — a powerful way to validate your model's performance.

## 🔍 Additional Notes & Advanced Usage

While the GUI is user-friendly, DeepLabCut also offers full control via the terminal with Python commands such as:

```python
import deeplabcut
deeplabcut.create_new_project(...)
deeplabcut.label_frames(config_path)
deeplabcut.train_network(config_path)

Your project folder is organized into subdirectories such as:

 - labeled-data/: Stores manually labeled frames

 - training-datasets/: Contains processed training/test datasets

 - videos/: Original or linked video files

 - dlc-models/ or dlc-models-pytorch/: Trained models and snapshots

-Post-Processing Tools
To improve results, DeepLabCut offers additional tools:

-Filtering predictions:
Use deeplabcut.filterpredictions() to apply smoothing (median or ARIMA/SARIMAX) to noisy predictions. This improves stability in downstream visualizations and tracking.

-Plotting trajectories:
Use deeplabcut.plot_trajectories(config_path, video_list) to visualize the keypoints’ motion across time (X/Y positions, likelihood, etc.).

-Label refinement (Active Learning):
Use deeplabcut.extract_outlier_frames() followed by deeplabcut.refine_labels() to fix poor predictions and re-train the model using updated annotations.

💡 These tools are helpful when working with long videos, low-confidence predictions, or rare behavioral patterns.


📚 For more advanced functions and explanations, see the [official DeepLabCut User Guide](https://deeplabcut.github.io/DeepLabCut/docs/standardDeepLabCut_UserGuide.html).
